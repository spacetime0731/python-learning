```python
def evaluate(func,x):
    return func(x)
def square(x):
    return x ** 2
def cube(x):
    return x ** 3
print(evaluate(square,5))
print(evaluate(cube,5))

```

```python
import math
def bisection(f, a, b, tol=1e-6):
    if f(a) * f(b) >= 0:
        print("错误：区间两端函数值同号，无法保证有根")
        return None

    count = 0
    while b - a > tol:
        mid = (a + b) / 2
        if f(a) * f(mid) <= 0:
            b = mid
        else:
            a = mid
        count += 1
    return (a + b) / 2, count
def g(x):
    return x ** 3 - x - 2
root, count = bisection(g, 1, 2)
theory = math.log2(1 / 1e-6)
print(f"根 ≈ {root:.6f}")
print(f"理论下界：{theory:.2f}")
print(f"验证：g(root) = {g(root):.2e}")
print(f"实际迭代：{count} 次")
print(f"理论取整：{math.ceil(theory)} 次")
    
            
```
