```python
import random
def estimate_pi(n):  # 撒 n 个点估算 π
    hits = 0
    for i in range(n):
        x = random.random()
        y = random.random()
        if x ** 2 + y ** 2 <=1:
            hits += 1
    return 4 * hits  / n # 落在扇形里的概率 = 扇形面积 / 正方形面积 = (π/4) / 1 = π/4
for n in [100, 10000, 1000000]:
    print(f"撒 {n:>9,} 个点：π ≈ {estimate_pi(n):.6f}") 
```
