```python
scores = [88,76,95,90]
credits = [5,3,1,4]
print(scores[0])
print(scores[-1])
print(len(scores))
print(scores[1:3])
```

```python
for score in scores:
    print(score)
```

```python
for i in range(4):
    print(f"第 {i+1} 门课:{scores[i]}分")
```

```python
weighted_sum = 0
total_credit = 0
for i in range(len(scores)):
    weighted_sum += credits[i] * scores[i]
    total_credit += credits[i]
gpa = weighted_sum / total_credit
print(f"加权平均数：{gpa:.2f}")
```
