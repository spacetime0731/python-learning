```python
text = "the quick brown fox jumps over the lazy dog the fox"
counts = {}
for word in text.split():
    if word in counts:
        counts[word] = counts[word] + 1
    else:
        counts[word] = 1
print(counts)
```

```python
counts2 = {}
for word in text.split():
    counts2[word] = counts2.get(word, 0) + 1
print(counts2)
```
