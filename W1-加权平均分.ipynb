```python
name="数学系短跑运动员"
pb=52.8
target=51.0
print(f"{name}当前PB：{pb}秒")
print(f"距离目标还差{pb-target:.1f}秒")
```

```python
math_credit = 5
math_score = 89
english_credit = 3
english_score = 76
pe_credit = 1
pe_score = 95
computer_credit = 4
computer_score=90
total_credit = math_credit + english_credit + pe_credit + computer_credit
weighted_sum = (math_credit * math_score + english_credit * english_score + pe_credit * pe_score + computer_credit * computer_score)
gpa=weighted_sum / total_credit
print(f"总学分：{total_credit}")
print(f"加权平均分:{gpa:.2f}")
```
